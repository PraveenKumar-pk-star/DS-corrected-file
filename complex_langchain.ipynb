```python
from langchain_core.prompts import ChatPromptTemplate,ChatMessagePromptTemplate,PromptTemplate
from langchain.chains import LLMChain
from langchain_openai import OpenAI
chatgpt_model=OpenAI(model="gpt-4o-mini",temperature=0.2)

promt_inpt1="{query}"
prompt_template=ChatPromptTemplate.from_template(promt_inpt1)
LLm_model=prompt_template | chatgpt_model
response=LLm_model.invoke({"query":"what is data in 2 lines"})
print(response)
```

```python
promt_input2="""tell story about the give {topics}."""
topic=["cat", "dog","elephant"]
prompt_template1=ChatPromptTemplate.from_template(promt_input2)
response1=[prompt_template1.format_messages(topics=user_input)for user_input in topic]
response1
```

```python
ans = [chatgpt_model.invoke(message) for message in response1]
ans
```
